```python
import torch
import torch.nn as nn
from transformers import AutoTokenizer, AutoModelForCausalLM, AutoConfig
import json
import numpy as np
from tqdm import tqdm
from google.colab import drive
import os
import gc

# --- 0. Mount Google Drive ---
if not os.path.exists('/content/drive'):
    drive.mount('/content/drive')

# --- 1. Hardware Precision Override ---
torch.backends.cuda.matmul.allow_tf32 = False
torch.backends.cudnn.allow_tf32 = False
device = "cuda" if torch.cuda.is_available() else "cpu"

# --- 2. Trimmed Architecture ---
class HybridInterpretabilityLens(nn.Module):
    def __init__(self, hidden_size=2560):
        super().__init__()
        self.thesis_adapter = nn.Sequential(
            nn.Linear(hidden_size, hidden_size // 2),
            nn.GELU(),
            nn.Linear(hidden_size // 2, hidden_size),
            nn.Dropout(0.1)
        )
        self.antithesis_scorer = nn.Sequential(
            nn.Linear(hidden_size, hidden_size // 4),
            nn.GELU(),
            nn.Linear(hidden_size // 4, 1),
            nn.Sigmoid()
        )
        self.anstoss_scorer = nn.Sequential(
            nn.Linear(hidden_size, hidden_size // 4),
            nn.GELU(),
            nn.Linear(hidden_size // 4, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        x = x.to(self.thesis_adapter[0].weight.dtype)
        p_ctx = self.thesis_adapter(x)
        contradiction = self.antithesis_scorer(x)
        impossibility = self.anstoss_scorer(x)
        friction_scalar = (contradiction + impossibility) / 2.0
        n_ctx = friction_scalar * x
        logic_vector = p_ctx - n_ctx
        return {"friction_scalar": friction_scalar, "negative_context": n_ctx, "logic_vector": logic_vector}

# --- 3. Loading Logic (Memory Optimized) ---
print("Cleaning GPU memory...")
torch.cuda.empty_cache()
gc.collect()

tokenizer = AutoTokenizer.from_pretrained("microsoft/phi-2")
if tokenizer.pad_token is None: tokenizer.pad_token = tokenizer.eos_token

BASE_CHECKPOINT = "/content/drive/MyDrive/Aufheben/checkpoints/hegelian_step_300_continued.pt"

if 'model' not in globals():
    print("Loading Base Model...")
    config = AutoConfig.from_pretrained("microsoft/phi-2")
    model = AutoModelForCausalLM.from_pretrained("microsoft/phi-2", config=config, torch_dtype=torch.float16).to(device)

try:
    checkpoint = torch.load(BASE_CHECKPOINT, map_location='cpu')
    sd = checkpoint['model_state_dict'] if 'model_state_dict' in checkpoint else checkpoint
    new_sd = {k.replace("base.model.", "model.").replace("base.", ""): v for k, v in sd.items()}
    model.load_state_dict(new_sd, strict=False)
    lens = HybridInterpretabilityLens(hidden_size=2560).to(device)
    lens_sd = lens.state_dict()
    for k in sd:
        if k.startswith(("thesis_adapter.", "antithesis_scorer.", "anstoss_scorer.")) and k in lens_sd:
            lens_sd[k] = sd[k]
    lens.load_state_dict(lens_sd)
    del checkpoint, sd, new_sd, lens_sd
    gc.collect()
    torch.cuda.empty_cache()
    print("✅ Weights Loaded and Memory Purged.")
except Exception as e:
    print(f"❌ Load failed: {e}")

lens.float().eval()
model.eval()
```

```python
# Optimized TruthfulQA loop using the Fast KV-Cache generator
INPUT_FILE = "/content/drive/MyDrive/truthfulqa_hallucination_poc (2).json"
OUTPUT_FILE = "/content/drive/MyDrive/point_hallucination_poc.json"

try:
    with open(INPUT_FILE, "r") as f:
        local_tqa_data = json.load(f)
except FileNotFoundError:
    print(f"Error: {INPUT_FILE} not found.")
    local_tqa_data = []

tqa_results = []

print(f"Running FAST generation on {len(local_tqa_data)} TruthfulQA prompts...")

for item in tqdm(local_tqa_data):
    prompt_data = {
        "prompt": item["prompt"],
        "id": item.get("id", "TQA-unknown"),
        "label": "TruthfulQA"
    }

    # Using the fast version with KV Caching
    metrics = generate_with_hybrid_lens_fast(
        prompt_data,
        model,
        lens,
        tokenizer,
        base_scale=1.5,
        max_new_tokens=60
    )

    metrics['suite'] = 'truthful_qa'
    tqa_results.append(metrics)

with open(OUTPUT_FILE, "w") as f:
    json.dump(tqa_results, f, indent=4)

print(f"\nDiagnostic complete. Results saved to: {OUTPUT_FILE}")
```

```python
def generate_with_hybrid_lens_fast(prompt_data, model, lens, tokenizer, base_scale=2.0, max_new_tokens=60):
    prompt = prompt_data["prompt"]
    inputs = tokenizer(prompt, return_tensors="pt").to(device)
    logic_norm_history, friction_scores = [], []

    # Get initial diagnostic
    with torch.no_grad():
        hidden = model.model(**inputs).last_hidden_state[:, -1:, :]
        initial_lens_out = lens(hidden.float())
        initial_friction = initial_lens_out["friction_scalar"].item()

    current_ids = inputs['input_ids']
    attention_mask = inputs['attention_mask']
    past_key_values = None
    input_len = current_ids.shape[1]

    for _ in range(max_new_tokens):
        with torch.no_grad():
            model_inputs = current_ids if past_key_values is None else current_ids[:, -1:]

            outputs = model.model(
                input_ids=model_inputs,
                attention_mask=attention_mask,
                past_key_values=past_key_values,
                use_cache=True
            )

            past_key_values = outputs.past_key_values
            last_hidden = outputs.last_hidden_state[:, -1:, :]

            lens_outputs = lens(last_hidden.float())
            logic_vector = lens_outputs["logic_vector"]
            friction = lens_outputs["friction_scalar"].item()

            logic_norm_history.append(logic_vector.norm().item())
            friction_scores.append(friction)

            dynamic_lambda = base_scale * (friction ** 2)
            steered_hidden = (last_hidden.float() + dynamic_lambda * logic_vector).to(model.dtype)

            logits = model.lm_head(steered_hidden)
            next_token_id = torch.argmax(logits, dim=-1)

            current_ids = torch.cat([current_ids, next_token_id], dim=-1)
            attention_mask = torch.cat([attention_mask, torch.ones((1, 1), device=device)], dim=-1)

            if next_token_id == tokenizer.eos_token_id:
                break

    return {
        "prompt": prompt,
        "label": prompt_data.get("label", "unknown"),
        "id": prompt_data.get("id", "unknown"),
        "response": tokenizer.decode(current_ids[0][input_len:], skip_special_tokens=True),
        "logic_norm_history": logic_norm_history,
        "friction_scores": friction_scores,
        "initial_friction": initial_friction,
        "peak_norm": max(logic_norm_history) if logic_norm_history else 0.0,
        "mean_friction": float(np.mean(friction_scores)) if friction_scores else 0.0
    }
```

```python
import json
import numpy as np
import pandas as pd
from collections import defaultdict

# Load your TruthfulQA results
with open("/content/drive/MyDrive/point_hallucination_poc.json") as f:
    results = json.load(f)

data = []

for item in results:
    prompt = item.get("prompt", "")
    response = item.get("response", "")
    label = item.get("label", "unknown")

    friction_scores = item.get("friction_scores", [])
    logic_norms = item.get("logic_norm_history", [])
    initial_friction = item.get("initial_friction", 0.0)
    peak_norm = item.get("peak_norm", 0.0)
    mean_friction = item.get("mean_friction", 0.0)

    # Find peak interference position
    peak_idx = np.argmax(friction_scores) if friction_scores else 0
    peak_token = "N/A"
    if peak_idx < len(response.split()):
        peak_token = response.split()[peak_idx:peak_idx+5]  # rough 5-word window

    data.append({
        "prompt": prompt[:150] + "..." if len(prompt) > 150 else prompt,
        "label": label,
        "initial_friction": round(initial_friction, 4),
        "mean_friction": round(mean_friction, 4),
        "peak_friction": round(max(friction_scores) if friction_scores else 0, 4),
        "peak_norm": round(peak_norm, 4),
        "peak_position": peak_idx,
        "peak_snippet": " ".join(peak_token),
        "response_length": len(response.split())
    })

df = pd.DataFrame(data)

print("=== TRUTHFULQA STEERING ANALYSIS ===")
print(df.describe())

# Category / Label breakdown
print("\n=== Per-Label Statistics ===")
print(df.groupby("label")[["mean_friction", "peak_friction", "peak_norm"]].mean().round(4))

# Strongest interference examples
print("\n=== Top 10 Strongest Interference Examples ===")
top = df.nlargest(10, "peak_friction")
print(top[["label", "mean_friction", "peak_friction", "peak_norm", "peak_snippet"]])

# Save detailed report
df.to_csv("truthfulqa_steering_analysis.csv", index=False)
print("\nFull analysis saved to truthfulqa_steering_analysis.csv")
```

```python
import torch
import numpy as np
import matplotlib.pyplot as plt
import umap.umap_ as umap
import json
import pandas as pd
import os

def visualize_anstoss_displacement(prompt, prompt_id, model, lens, tokenizer, save_dir, base_scale=1.5, max_new_tokens=20):
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    original_states, steered_states, tokens, frictions = [], [], [], []
    current_ids, attention_mask, past_key_values = inputs['input_ids'], inputs['attention_mask'], None

    print(f"\nTracking latent trajectory for [{prompt_id}]: '{prompt[:50]}...'")

    for step in range(max_new_tokens):
        with torch.no_grad():
            model_inputs = current_ids if past_key_values is None else current_ids[:, -1:]
            outputs = model.model(input_ids=model_inputs, attention_mask=attention_mask, past_key_values=past_key_values, use_cache=True)
            past_key_values = outputs.past_key_values
            last_hidden = outputs.last_hidden_state[:, -1:, :]
            original_states.append(last_hidden.squeeze().cpu().numpy())

            lens_outputs = lens(last_hidden.float())
            logic_vector, friction = lens_outputs["logic_vector"], lens_outputs["friction_scalar"].item()
            frictions.append(friction)

            steered_hidden = (last_hidden.float() + (base_scale * (friction ** 2)) * logic_vector).to(model.dtype)
            steered_states.append(steered_hidden.squeeze().cpu().numpy())

            logits = model.lm_head(steered_hidden)
            next_token_id = torch.argmax(logits, dim=-1)
            current_ids = torch.cat([current_ids, next_token_id], dim=-1)
            attention_mask = torch.cat([attention_mask, torch.ones((1, 1), device=model.device)], dim=-1)
            tokens.append(tokenizer.decode(next_token_id))
            if next_token_id == tokenizer.eos_token_id: break

    all_states = np.vstack(original_states + steered_states)
    reducer = umap.UMAP(n_neighbors=min(5, len(tokens)), min_dist=0.3, metric='cosine', random_state=42)
    embedding = reducer.fit_transform(all_states)
    orig_2d, steer_2d = embedding[:len(tokens)], embedding[len(tokens):]

    plt.figure(figsize=(10, 6))
    plt.style.use('dark_background')
    plt.plot(orig_2d[:, 0], orig_2d[:, 1], color='gray', alpha=0.4, linestyle='--', label="Base LLM")
    plt.plot(steer_2d[:, 0], steer_2d[:, 1], color='cyan', alpha=0.8, linewidth=2, label="Steered")

    for i in range(len(tokens)):
        color = plt.cm.YlOrRd(frictions[i])
        plt.arrow(orig_2d[i, 0], orig_2d[i, 1], steer_2d[i, 0] - orig_2d[i, 0], steer_2d[i, 1] - orig_2d[i, 1], color=color, alpha=0.6, length_includes_head=True, head_width=0.08)
        plt.text(steer_2d[i, 0], steer_2d[i, 1] + 0.05, tokens[i], color='cyan', fontsize=8)

    plt.title(f"Manifold Displacement: {prompt_id}\n{prompt[:60]}")
    plt.legend()
    save_path = os.path.join(save_dir, f"{prompt_id.replace('-', '_')}_manifold.png")
    plt.savefig(save_path, dpi=200)
    plt.show()
    plt.close()

SAVE_DIR = "/content/drive/MyDrive/Aufheben/manifold_plots"
os.makedirs(SAVE_DIR, exist_ok=True)

try:
    with open("/content/drive/MyDrive/point_hallucination_poc.json", "r") as f:
        df = pd.DataFrame(json.load(f))
    top_20 = df.nlargest(20, 'peak_norm')
    print(f"Processing top 20 tension cases and saving to {SAVE_DIR}...")
    for _, row in top_20.iterrows():
        visualize_anstoss_displacement(row['prompt'], row['id'], model, lens, tokenizer, SAVE_DIR)
except Exception as e:
    print(f"Error: {e}")
```
